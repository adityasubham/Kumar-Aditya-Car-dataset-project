##  Check and Remove Duplicate Records

```python
#check if any duplicates
df.duplicated().sum()
```

The dataset contains 1,202 duplicate records. These records have the same values across all the columns and therefore we will not add any new information to the analysis. Keeping these repeated records could affect the accuracy of our results, so they will be removed before continuing with further analysis.

```python
#to remove all duplicates
df.drop_duplicates(inplace=True)
```

```python
#check if any duplicates
df.duplicated().sum()
```

**Observation:**

After removing the duplicate records, the dataset contains 6,926 unique records and 10 columns. This gives us a cleaner dataset without repeated entries, which will help make the upcoming analysis more reliable.

##  Handle Missing Values

```python
#check missing value
df.isnull().sum()
```

```python
#checking missing values >0
missing = df.isnull().sum()
missing[missing > 0]
```

The dataset contains missing values in two columns: mileage and engine. Both columns have 208 missing values out of 8,128 records. The remaining columns have complete data. Since mileage and engine capacity are important features for analyzing used cars, these missing values need to be handled properly before continuing with the analysis.

```python
df['mileage'] = df['mileage'].fillna(df['mileage'].mode())
```

```python
df['engine'] = df['engine'].fillna(df['engine'].mode())
```

**Observation:**

The dataset had missing values in the engine and mileage columns. Each column contained 208 missing values. Since both attributes are important for car analysis, the missing values were handled using the mode, which is the most frequently occurring value in the respective column. This helped retain the available records without removing rows from the dataset.

```python
#select only the numerical columns from the car dataset
numeric_cols = df.select_dtypes(include = "number")
print(numeric_cols)
```

The dataset contains 3 numerical columns: year, selling price, and kilometers driven. These columns will be useful for understanding car prices, vehicle age, and usage patterns during the analysis.

```python
#to find and visualize outliers in the numerical columns
numeric_cols.boxplot(figsize =(12,6))
plt.xticks(rotation = 90)
plt.show()
```

The box plot was used to visually identify unusual values in the numerical columns. It helps us understand the spread of the data and detect possible outliers in car prices, manufacturing year, and kilometers driven.

```python
# Distribution of Selling Price
plt.figure(figsize=(8,5))
plt.hist(df["selling_price"], bins=20)
plt.title("Distribution of Car Selling Prices")
plt.xlabel("Selling Price")
plt.ylabel("Number of Cars")

plt.show()
```

**Observation:**

The selling price of the cars varies widely across the dataset. Most cars are available at lower to mid-range prices, while only a few cars have very high prices. This shows that the selling price is not evenly distributed and contains some higher-priced cars.

```python
# Distribution of Fuel Types
fuel_counts = df[["fuel_CNG", "fuel_Diesel", "fuel_LPG", "fuel_Petrol"]].sum()

fuel_counts.plot(kind="bar", figsize=(8,5))

plt.title("Distribution of Fuel Types")
plt.xlabel("Fuel Type")
plt.ylabel("Number of Cars")
plt.xticks(rotation=0)

plt.show()
```

The dataset contains cars using different fuel types, with fuel_Diesel categories being much more than others.

```python
# Distribution of Transmission Types
transmission_counts = df["transmission"].value_counts().sort_index()

transmission_counts.plot(kind="bar", figsize=(7,5))

plt.title("Distribution of Transmission Types")
plt.xlabel("Transmission Type")
plt.ylabel("Number of Cars")
plt.xticks([0, 1], ["Manual", "Automatic"], rotation=0)

plt.show()
```

The dataset contains both manual and automatic cars, with manual type being more common than the automatic.

```python
# Box Plot of Kilometers Driven
plt.figure(figsize=(8,5))

plt.boxplot(df["km_driven"])

plt.title("Box Plot of Kilometers Driven")
plt.ylabel("Kilometers Driven")

plt.show()
```

Observation:

The box plot shows the spread of kilometers driven and highlights some cars with unusually high driving distances.

```python
# Selling Price vs Manufacturing Year
plt.figure(figsize=(8,5))

plt.scatter(df["year"], df["selling_price"], alpha=0.5)

plt.title("Selling Price vs Manufacturing Year")
plt.xlabel("Manufacturing Year")
plt.ylabel("Selling Price")

plt.show()
```

The plot shows that newer cars generally tend to have higher selling prices, although the price also varies between cars of the same manufacturing year.

```python
# Bivariate
import matplotlib.pyplot as plt
plt.scatter(df["selling_price"], df["year"])
plt.title("Selling Price vs Year")
plt.xlabel("Selling Price")
plt.ylabel("Year")
plt.show()
```

**Observation:**

The graph shows that newer cars generally have higher selling prices compared to older cars. However, the price is not decided by the year alone, as factors like mileage, fuel type, brand, and car condition also affect the price.

```python
# Multivariate Analysis
sns.pairplot(df[["year", "selling_price", "km_driven", "mileage", "engine"]])
plt.show()
```

**Observation:**


The pair plot helps us understand the relationships between different car features at the same time. It shows that factors such as the car's year, selling price, mileage, engine size, and kilometers driven are related in different ways. This gives us a better overall understanding of the dataset.

```python
# HeatMap
corr = df.corr(numeric_only = True)
plt.figure(figsize= (10,6))
sns.heatmap(corr, annot = True, cmap = "coolwarm")
plt.title("Correlation of variables")
plt.show()
```

**Observation:**

The heatmap shows the correlation between the numerical variables in the dataset. It helps us understand which variables are more closely related to each other. The correlation between selling price and year shows that newer cars generally have higher prices, while the relationship with kilometers driven is comparatively weaker.
